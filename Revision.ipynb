```python
#Goal: Write a function that takes a list of passwords (strings).
```

```python
def check_security(passwords: list[str]):
    valid_password = []
    for current_word in passwords:
        if current_word == "Hacker":
            return "Alarm!"
        elif len(current_word) >= 8:
            valid_password.append(current_word)
    return valid_password

good_list = ["apple123", "cat", "supersecret99"]
bad_list = ["dog12345", "HACKER", "elephant"]

print(check_security(good_list)) 
print(check_security(bad_list))
```

```python
def check_cart(prices: list[int]):
    expensive_items = []
    for current_price in prices:
        if current_price == 0:
            return "Security!"
        elif current_price > 50:
            expensive_items.append(current_price)
    return expensive_items
        
normal_cart = [20, 99, 15, 100]
thief_cart = [10, 5, 0, 80]

print(check_cart(normal_cart)) 
print(check_cart(thief_cart))  
```
