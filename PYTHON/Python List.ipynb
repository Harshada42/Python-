```python
#Coding question
#Find maximum value in a list
numbers = [34,56,78,95,23]
maximum = numbers[0]
for num in numbers:
    if num > maximum:
        maximum = num
print("Maximum value is:", maximum)
```

```python
#Find minimum value in list
numbers = [45,65,13,64,12]
minimum = numbers[0]
for num in numbers:
    if num < minimum:
        minimum = num
print("minimum value is:", minimum)
```

```python
#Find even and odd number
numbers = [42,78,35,53,12]
even_number = []
odd_number = []

for num in numbers:
    if num % 2 == 0:
        even_number.append(num)
    else:
        odd_number.append(num)

print("Even numbers:", even_number)
print("Odd numbers:", odd_number)
```

```python
#Finding Even numbers
numbers = [42,78,35,53,12]

even_numbers = [num for num in numbers if num % 2 == 0]

print(even_numbers)
```

```python
#Finding odd numbers
numbers = [42,78,35,53,12]

odd_numbers = [num for num in numbers if num % 2 != 0]

print(odd_numbers)
```
