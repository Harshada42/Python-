```python
#Looping through dictionaries
#Loop through keys
car = {"brand": "Tesla", "model": "Model 3", "year": 2024}
for key in car.keys():
    print(key)
```

```python
#Loop through values
for value in car.values():
    print(value)
```

```python
# loop through both
for key, value in car.items():
    print(f"{key}: {value}")
```

```python
#Count frequency of items
items = ["A", "B", "C", "D","D", "B", "C", "D"]

frequency = {}

for item in items:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1
print(frequency)
```

```python
names = ["alisha", "anna", "kamila", "sasha", "alisha"]

howmany = {}

for name in names:
    if name in howmany:
        howmany[name] += 1
    else:
        howmany[name] = 1

print(howmany)
```
