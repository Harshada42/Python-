```python
#Goal: Write a function that checks 
#how fast a car is driving (in miles per hour) and returns the correct action.
def check_speed(speed: int) -> str:
    if speed > 70:
        return "Ticket"
    elif speed >= 50:
        return "Warning"
    else:
        return "Safe"
print(check_speed(80))
print(check_speed(60))
print(check_speed(40))
```

```python
#Goal: Write a function that looks at a 
#player's health points (HP) and returns their status.
def check_health(hp: int) -> str:
    if hp == 0:
        return "Game Over"
    elif hp < 20:
        return "Critical"
    else:
        return "Healthy"

print(check_health(100)) # Should print "Healthy"
print(check_health(15))  # Should print "Critical"
print(check_health(0))   # Should print "Game Over"
```

```python
#Goal: Write a function that checks if a user typed
#the correct password. The secret password is the word "OpenSesame".
def login(password: str) -> str:
    if password == "OpenSesame":
        return "Access Granted"
    else:
        return "Access Denied"
print(login("OpenSesame"))
print(login("12345"))
```

```python
#Goal: Loop through a list of words. Every time you see the word "apple",
#add 1 to your score. Return the final score.
def count_apples(words: list[str]) -> int:
    score = 0
    for word in words:
        if word == "apple":
            score = score + 1
    return score


print(count_apples(["apple", "Grape", "apple", "banana"]))
```

```python
#Goal: You are given a list of words. Create a brand new, 
#empty list. Loop through the words. If a word is exactly 3 letters long, 
#shove it into your new list. Hand back the new list.
def count_total_letters(words: list[str]) -> int:
    score = 0
    for word in words:
        len_of_word = len(word)
        score = score + len_of_word
    return score
print(count_total_letters(["hi","cat","dog"]))
```
