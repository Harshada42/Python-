```python
# Practice Mission: If else elif
#Write a function called player_status(health).

#If health is exactly 0, return "Game Over".

#If health is less than 30, return "Low Health!".

#Otherwise, return "Ready to fight".
```

```python
def player_status(health):
    if health == 0:
        return "Game Over"
    elif health < 30:
        return "Low Health"
    else:
        return "Ready to fight"
print(player_status(15))
```

```python
#2 Practice Mission: For Loop
#You have a list of damage numbers
#from your attacks: [10, 20, 50]. You need to calculate the total damage.
```

```python
def calculate_damage(attack_list):
    total = 0
    for attack in attack_list:
        total = total + attack
    return total
print(calculate_damage([10,20,50]))
```

```python
# Your Practice Mission:
#Your player starts with 50 stamina. 
#Every time they swing their sword, it costs 15 stamina.
```

```python
def swing_sword():
    stamina = 50
    swings = 0
    while stamina >= 15:
        stamina = stamina - 15
        swings = swings + 1
    return swings
print(swing_sword())
```
