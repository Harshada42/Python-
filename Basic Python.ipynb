```python
def check_grades(scores: list[int]) -> list[str] :
    result = []
    for score in scores:
        if score >= 97:
            result.append("A")
        elif score >= 55:
            result.append("B")
        elif score >= 35:
            result.append("C")
        else:  
            result.append("Needs Improvement")
    return result

my_test_score = [92, 56, 45]
final_report = check_grades(my_test_score)
print(final_report)
```

```python
def check_weather(temp: int) -> str:
    if temp > 30:
        return "Hot"
    elif temp < 10:
        return "Cold"
    else:
        return "Perfect"



# Test it out!
print(check_weather(35)) # Should print "Hot"
print(check_weather(20)) # Should print "Perfect"
print(check_weather(5))  # Should print "Cold"
```

```python
def get_ticket_type(age: int) -> str:
    if age > 13:
        return "Child"
    elif age >= 65:
        return "Senior"
    else:
        return "Adult"




print(get_ticket_type(10)) 
print(get_ticket_type(30)) 
print(get_ticket_type(70)) 
```
