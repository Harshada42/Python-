```python
#Goal: The Candy Factory
#You are given a list of candy strings.

#Setup: Create an empty dictionary for tally marks.

#Tour Guide: Loop through the list of candies.

#Bouncer:

#If you see the candy "poison", immediately hit the eject button and return "Shut down!".

#Otherwise, add the candy to your dictionary and update its tally marks (just like we did for the Most Frequent problem!).

#Delivery: Return the dictionary of tally marks at the very end.


def count_candy(candies: list[str]):
    # 1. SETUP
    tally = {}
    for candy in candies:
        if candy == "poison":
            return "Shut Down!"
        elif candy not in tally:
            tally[candy] = 1
        else:
            tally[candy] = tally[candy] + 1
    return tally
      

print(count_candy(["skittle", "m&m", "skittle"])) 
print(count_candy(["skittle", "poison", "m&m"]))  
```

```python
#Goal: You are given a list of guest ages at a club. 
#I want you to collect all the ages that are 18 or older and
#hand them back in a new collection.

#Hint: Ask yourself, do you just need to collect items in a line,
#or do you need to connect labels to values?

def collect_adults(ages: list[int]):
    # 1. SETUP: Create your empty List or Dictionary
    adult = []
    for age in ages:
        if age >= 18:
            adult.append(age)
    return adult
    
print(collect_adults([16, 21, 17, 30])) 
```

```python
def is_palindrome(nums: list[int]):
    # 1. SETUP: Create our two pointers
    left = 0
    right = len(nums) - 1
    
    # 2. TOUR GUIDE: Keep looping until they meet in the middle
    while left < right:
        if nums[left] != nums[right]:
            return False
        
        
        # Move the pointers toward the middle so we don't get stuck forever!
        left = left + 1
        right = right - 1
        
    # 4. DELIVERY: If we survived the loop, it's a perfect palindrome!
    return True

# Test it out!
good_list = [1, 2, 3, 2, 1]
bad_list = [1, 2, 3, 4, 5]

print(is_palindrome(good_list)) # Should print True
print(is_palindrome(bad_list))  # Should print False
```

```python
def launch_rocket(countdown: int):
    # Keep looping AS LONG AS the countdown is greater than 0
    while countdown > 0:
        print(countdown)
        
        # THE MOST IMPORTANT LINE: 
        # If we forget to subtract 1, the computer will print forever!
        countdown = countdown - 1
        
    return "Liftoff!"

print(launch_rocket(3))
```

```python
def chop_to_one(number: int):
    chops = 0
    
    # Keep looping AS LONG AS the number is greater than 1
    while number > 1:
        # Divide the number in half (the // means divide without decimals)
        number = number // 2 
        
        # Add a tally mark to our chop count
        chops = chops + 1
        
    return chops

print(chop_to_one(32)) # Takes exactly 5 chops to reach 1!
```

```python
def monitor_battery():
    battery = 50
    while battery > 0:
        if battery <= 20:
            return "Low Battery!"
        battery = battery - 10
    return "Phone Died"
            

    

# Test it out!
print(monitor_battery())
```

```python
def pair_sum_brute_force(nums: list[int], target: int):
    
    # 1. OUTER LOOP: Pick the first number (the "left" hand)
    for i in range(len(nums)):
        
        # 2. INNER LOOP: Pick the second number (the "right" hand)
        # We start at 'i + 1' so we don't add a number to itself!
        for j in range(i + 1, len(nums)):
            
            # 3. BOUNCER: Check if these two specific seats add up!
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
                
    # 4. DELIVERY
    return "No pair found"

print(pair_sum_brute_force([2, 7, 11, 15], 9)) # Prints [2, 7]
```
