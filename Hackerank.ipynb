```python
#Practice 1
N = 2
A = [10,30,50,20]
B = [20, 10, 40, 30]
```

```python
total = 0
```

```python
differences = []
```

```python
for i in range(2 * N):

    city_b = min(A[i], B[i]) + B[i]

    total += city_b

    difference = A[i] - city_b

    differences.append(difference)

differences.sort()

total += sum(differences[:N])

print(total)
    
```

```python
N = 2

A = [15, 40, 25, 35]
B = [20, 30, 10, 50]

total = 0
differences = []

for i in range(2 * N):
    city_b = min(A[i], B[i]) + B[i]
    total += city_b

    difference = A[i] - city_b
    differences.append(difference)

differences.sort()

total += sum(differences[:N])

print(total)    
```
